```python
#old data will be deleted and new data will be written to the file

with open("dummy,txt", "w") as f:
    f.write("This is a dummy file")
    f.close()
```

```python
print(os.listdir())

with open("dummy.txt", "r") as f:
    print(f.read())
    f.close()
```

```python
#next mode will be append mode, which will add new data to the end of the file without deleting the old data
# add more data to the file without deleting the old data

with open("dummy.txt", "a") as f:
    f.write("\nThis is a new line added to the file")
    print(f.tell())
```

```python
# "r+" is a mode that allows you to read and write to a file. It opens the file for both reading and writing. 
# The file pointer is placed at the beginning of the file. If the file does not exist, it will raise an error.
# it will not wipe out the old data, but it will allow you to read and write to the file.

with open("dummy.txt", "r+") as f:
    print(f.read())
    f.write("\nThis is a new line added to the file using r+ mode")
    print(f.tell())
    print(f.read())
```
