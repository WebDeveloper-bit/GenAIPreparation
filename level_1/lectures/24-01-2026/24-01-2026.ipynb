# STRING METHODS

```python
# string basics demo (use unique variable names to avoid overwriting existing notebook variables)

str_demo = "Hello, World!"
multi_line = """This is a
multi-line string."""
raw_path = r"C:\new_folder\test.txt"

print("Original:", str_demo)
print("Length:", len(str_demo))
print("Index 7:", str_demo[7])
print("Slice [0:5]:", str_demo[:5])
print("Concat + and *:", str_demo + " :-) " + "Hi" * 2)

# immutability: produce a new string instead of modifying in place
replaced = str_demo.replace("World", "Python")
print("Replaced:", replaced)

# common methods
print("Upper / Lower:", str_demo.upper(), "/", str_demo.lower())
print("Title:", "the quick brown fox".title())
print("Strip:", "  padded  ".strip())
print("Split:", "one,two,three".split(","))
print("Join:", "-".join(["a", "b", "c"]))
print("Find 'World':", str_demo.find("World"))
print("Starts/Endswith:", str_demo.startswith("Hello"), str_demo.endswith("!"))

# formatting
name = "Alice"; age = 30
print("format():", "Name: {}, Age: {}".format(name, age))
print("f-string:", f"Name: {name}, Age: {age}")

# raw and multi-line examples
print("Raw path:", raw_path)
print("Multi-line:\n", multi_line)
```

# Control Flow

```python
lst3 = [1,2,3,4,5] 

for i in lst3:
    if i == 1:
        continue
    print(i)
```

```python
count = 0 ##break statement is used to exit the loop when a certain condition is met. In this case, the loop will continue until count reaches 5, at which point it will break out of the loop.
while True:
    count += 1
    if count >= 5:
        break
    print("Count is:", count)
```
