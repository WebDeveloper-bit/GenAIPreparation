```python
#globle variable
a=0

def add():
    a=3 #local variable
    a+=1
    return a
print(a)
a=add()
print(a)
```

```python
#to use globale variable inside function we have to use "global" keyword

#globle variable
a=0

def add():
    global a
    a+=1
    return a
print(a)
a=add()
print(a)
```

```python
#if there are multiple values passed to function then we can use *args and **kwargs

def sum(*args):
    print(type(args))
    sum =0
    for i in args:
        sum+=i
    print("Sum is:",sum)
sum(1,2,3,4,5)
```

```python
#if there are multiple values passed to function then we can use *args and **kwargs

def sum(a,b,*args):
    print(type(a))
    print(type(b))
    print(type(args))
    sum =0
    for i in args:
        sum+=i
    print("Sum is:",sum)
sum(1,2,3,4,5)
```

```python
def person(**kwargs):
    print(type(kwargs))
    for key, value in kwargs.items():
        print(f"{key}: {value}")

person(name="John", age=30, city="New York")
```

```python
def order_pizza(size, *toppings, **details):
    print(f"Ordering a {size} pizza with the following toppings:")
    for topping in toppings:
        print(f"- {topping}")
    if details:
        print("Additional details:")
        for key, value in details.items():
            print(f"- {key}: {value}")
```

```python
order_pizza(4, "large", "pepperoni", "mushrooms", crust="thin", extra_cheese=True)
```
